```python
df = pd.read_excel(r'C:\Users\noura\OneDrive\Desktop\Data Cleaning\Customer Call List .xlsx')
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns="Not_Useful_Column")
```

```python
#df["Last_Name"] = df['Last_Name'].str.strip('...')
#df["Last_Name"] = df['Last_Name'].str.strip('/')
#df["Last_Name"] = df['Last_Name'].str.strip('_')
df['Last_Name'] = df['Last_Name'].astype(str).str.replace(r'[^a-zA-Za-z]', '', regex=True)
df['Last_Name'] = df['Last_Name'].str.strip()
df
```

```python
df['Phone_Number']=df['Phone_Number'].astype(str).str.replace(r'[^0-9]', '', regex= True)
df['Phone_Number'] = df['Phone_Number'].apply(lambda x :f'{str(x)[:3]}-{str(x)[3:6]}-{str(x)[6:10]}' if len(str(x)) ==10 else '')
df
```

```python
df[['Street_Address', 'State', 'Zip_Code']] = df['Address'].str.split(',', n=2 ,expand=True)
df.head()
```

```python
df['Paying Customer']=df['Paying Customer'].str.replace('Yes' ,'Y')
df['Paying Customer']=df['Paying Customer'].str.replace('No' ,'N')

df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Yes' ,'Y')
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('No' ,'N')
df
```

```python
df=df.fillna('')
```

```python
df.drop(df[df['Do_Not_Contact'] == 'Y'].index, inplace = True)
df.drop(df[df['Phone_Number'] == ''].index, inplace = True)
df
```

```python
df.reset_index(drop= True)
df
```
